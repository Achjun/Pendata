# Normalisasi Data Menggunakan sklearn dan Fungsi Sendiri

Notebook ini menunjukkan cara melakukan normalisasi data menggunakan library **scikit-learn (sklearn)** serta membuat fungsi normalisasi sendiri.

Dataset contoh menggunakan atribut **Usia**.

```python
import pandas as pd

data = {
    'ID':[1,2,3,4,5,6,7,8],
    'Usia':[18,25,20,21,19,22,23,24]
}

df = pd.DataFrame(data)
df
```

## Normalisasi Menggunakan sklearn

Metode yang digunakan:
- MinMaxScaler
- StandardScaler (Z-Score)

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

minmax = MinMaxScaler()
zscore = StandardScaler()

df['MinMax'] = minmax.fit_transform(df[['Usia']])
df['ZScore'] = zscore.fit_transform(df[['Usia']])

df
```

## Normalisasi Menggunakan Fungsi Sendiri

```python
def minmax_normalization(data):
    xmin = min(data)
    xmax = max(data)
    return [(x - xmin) / (xmax - xmin) for x in data]

usia = df['Usia'].tolist()

df['MinMax_manual'] = minmax_normalization(usia)

df
```

```python
import numpy as np

def zscore_normalization(data):
    mean = np.mean(data)
    std = np.std(data)
    return [(x - mean) / std for x in data]

df['ZScore_manual'] = zscore_normalization(usia)

df
```
